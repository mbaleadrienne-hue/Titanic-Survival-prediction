```python
## 1_ Import libraries
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix,ConfusionMatrixDisplay,classification_report
import matplotlib.pyplot as plt
```

```python
## 2_Import Important DataSet
```

```python
Df=pd.read_csv('train.csv')
```

```python
Df.head()
```

```python
Df.info()
```

```python
Df.isnull().sum()
```

```python
## 3_Splitting Features and Target
```

```python
x=Df.drop('Survived', axis=1)
y=Df['Survived']
```

```python
num_cols=['Age','Fare']
cat_cols=['Sex','Embarked']
```

```python
## 4_Splitting Train_Test_Split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
## 5_Imputation
```

```python
impute=SimpleImputer(strategy='median')
x_train[num_cols]=impute.fit_transform(x_train[num_cols])
x_test[num_cols]=impute.transform(x_test[num_cols])
```

```python
cat_impute=SimpleImputer(strategy='most_frequent')
x_train[cat_cols]=cat_impute.fit_transform(x_train[cat_cols])
x_test[cat_cols]=cat_impute.transform(x_test[cat_cols])
```

```python
## 6_Encode
```

```python
encoder=OneHotEncoder(handle_unknown='ignore',sparse_output=False)
x_train_encoded=encoder.fit_transform(x_train[cat_cols])
x_test_encoded=encoder.transform(x_test[cat_cols])
```

```python
new_cols=encoder.get_feature_names_out(cat_cols)
x_train_encoded=pd.DataFrame(x_train_encoded,columns=new_cols,index=x_train.index)
x_test_encoded=pd.DataFrame(x_test_encoded,columns=new_cols,index=x_test.index)
```

```python
x_train_final=pd.concat([x_train[num_cols],x_train_encoded],axis=1)
x_test_final=pd.concat([x_test[num_cols],x_test_encoded],axis=1)
```

```python
## 7_ Model LogisticRegression
```

```python
model=LogisticRegression(max_iter=1000)
```

```python
model.fit(x_train_final, y_train)
y_pred=model.predict(x_test_final)

print(f'Accuracy:{accuracy_score(y_test,y_pred):.2f}')
```

```python
## 8_confusion Matrix
```

```python
cm=confusion_matrix(y_test,y_pred)
print(cm)
```

```python
## Random Forest
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf= RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(x_train_final,y_train)

y_pred_rf=rf.predict(x_test_final)
```

```python
print('Accuracy RF:',accuracy_score(y_test,y_pred_rf))
print(confusion_matrix(y_test,y_pred_rf))
print(classification_report(y_test,y_pred_rf))
```
